```python
import numpy as np
from IPython.display import display
import pandas as pd
```

```python
df = pd.read_csv("C:/Users/Selin/Desktop/mobile (1).csv")
```

```python
# question num 1
def last_one(name):
    return str(name)[0]

df['First name'] = df['Customer Name'].apply(last_one)
df
```

```python
# question num 2
prefix_to_company = {
    '050': 'Pelephone',
    '051': 'Golan',
    '052': 'Cellcom',
    '053': 'Motorola'
}

df['Cell phone company'] = df['Phone Number'].str[:3]

df['Cell phone company'] = df['Cell phone company'].replace(prefix_to_company)
df
```

```python
# question num 3 A
df['Cell phone company'].value_counts()
```

```python
# question num 3 B
df['Cell phone company'].unique()
```

```python
# question num 3 C
df['Cell phone company'].nunique()
```

```python
# question num 4 A
df['User Type'] = df.apply(lambda row: 'talker' if row['Talked'] > row['Internet'] else 'surfer',axis=1)
df
```

```python
# question num 4 B
def user_type(call, internet):
    return 'talker' if call > internet else 'surfer'

df['User Typ'] = np.vectorize(user_type)(df['Talked'], df['Internet'])
```

```python
# question num 5 A
df_sorted_talk = df.sort_values('Talked', ascending=True)
df_sorted_talk
```

```python
# question num 5 B
df['Total Time'] = df['Talked'] + df['Internet']
df_sorted_by_total = df.sort_values(by='Total Time', ascending=True)
df_sorted_by_total
```

```python
# question num 6 A
top_5_talkers = df.nlargest(5, 'Talked')
top_5_talkers['Customer Name']
```

```python
# question num 6 B
bottom_3_surfers = df.nsmallest(3, 'Internet')
bottom_3_surfers['Customer Name']
```
